### comparing the cut of the diamond and the pricing 

```python
# grouping categorical cut by the prices
price_by_cut=df.groupby('cut',observed=False)['price'].agg(['max','min','mean','count']).sort_values('max',ascending=False)
price_by_cut.round(2)
```

The data shows that the highest price is not just about the quality of the cut, but also about how big (carat) and clear the diamond is. Therefore, even though "Ideal" sounds best, it's the combination of factors — especially carat size — that determines the top price.

### comparing the clarity and the pricing 

```python
#Clarity by price 
price_by_clarity=df.groupby('clarity',observed=False)['price'].agg(['max','min','mean','count']).sort_values('mean',ascending=False)
price_by_clarity.round(2)
```

Although SI2 is a lower clarity grade, diamonds with this clarity have the highest average price in the dataset — not because they are clearer, but because other attributes like carat (weight) and cut quality play a significant role in determining the price.

This shows that diamond pricing is multi-factorial, and clarity alone does not determine value.

```python
color_by_price=df.groupby('color',observed=False)['price'].agg(['max','min','mean','count']).sort_values('mean',ascending=False)
color_by_price.round(2)
```

Even though D diamonds are the clearest, that doesn’t mean they always cost the most.
Sometimes a diamond with a little color but a better shape or heavier weight can cost more.
So we look at everything together, not just the color

## 4.Detect outliers ,skewed distributions and multicolinearity

```python
# Finding Price Outliers
mean_value=df['price'].mean()
std_dev=df['price'].std()

lower_bound=mean_value-3 * std_dev
upper_bound=mean_value + 3 *std_dev

outliers=df[(df['price']<lower_bound) | (df['price']>upper_bound) ]
len(outliers)
```

```python
# finding Carat outliers
mean_value1 = df['carat'].mean()
std_dev1 = df['carat'].std()

lower_bound1 = mean_value1 - 3 * std_dev1
upper_bound1 = mean_value1 + 3 * std_dev1

outliers_1 = df[(df['carat'] < lower_bound1) | (df['carat'] > upper_bound1)]
len(outliers_1)
```

```python
# Finding table ouliers
mean_value2 = df['table'].mean()
std_dev2 = df['table'].std()

lower_bound2 = mean_value2 - 3 * std_dev2
upper_bound2 = mean_value2 + 3 * std_dev2

outliers_2 = df[(df['table'] < lower_bound2) | (df['table'] > upper_bound2)]
len(outliers_2)
```

```python
#Finding depth outliers
mean_value3 = df['depth'].mean()
std_dev3 = df['depth'].std()

lower_bound3 = mean_value3 - 3 * std_dev3
upper_bound3 = mean_value3 + 3 * std_dev3

outliers_3= df[(df['depth'] < lower_bound3) | (df['depth'] > upper_bound3)]
len(outliers_3)
```

```python
df.describe()
```

```python
# outlier removal 
Non_outliers_price = df[(df['price'] <= upper_bound) & (df['price'] >= lower_bound)]
Non_outliers_carat = df[(df['carat'] <= upper_bound1) & (df['carat'] >= lower_bound1)]
Non_outliers_depth = df[(df['depth'] <= upper_bound2) & (df['depth'] >= lower_bound2)]
Non_outliers_table = df[(df['table'] <= upper_bound3) & (df['table'] >= lower_bound3)]
```

```python
# Comparison of the price with and without outliers
fig, axes = plt.subplots(1, 2, figsize=(10, 4))  # 1 row, 2 columns
plt.suptitle("Plot 6 : Comparison of prices with and without outliers")

# Plot on each axis
sns.boxplot(y=Non_outliers_price['price'], ax=axes[0])
axes[0].set_title('Boxplot (Non-Outliers)')

sns.boxplot(data=df, y='price', ax=axes[1])
axes[1].set_title('Boxplot (With Outliers)')
plt.savefig(r"C:\Users\ADMIN\Documents\AFRICDSA\Python\Diamond market analysis\images\outliers_in_price_boxplot",bbox_inches='tight')

plt.tight_layout()
plt.show()
```

The boxplot comparison shows how outliers affect the distribution of diamond prices. When outliers are included, the plot becomes stretched, making it hard to see the main price range. After removing outliers, the boxplot clearly shows that most diamond prices are concentrated below $ 5,000. This helps us better understand the typical price distribution.
